# clean base

```python

def clean_text(text):
    if pd.isnull(text):
        return ""
    text = text.lower()
    # Giữ lại . ? ! để dùng cho tách câu
    text = re.sub(r'[^a-zA-ZÀ-ỹ0-9%\s\.\,\"]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
def sentence_segmentation(text):
    if not text:
        return []
    # Tách theo ., ?, ! hoặc xuống dòng
    sentences = re.split(r'[.\n]+', text)
    # Loại bỏ câu trống
    sentences = [s.strip() for s in sentences if s.strip()]
    return sentences
def normalize_sentence_case(sentence):
    sentence = sentence.strip()
    if not sentence:
        return ""
    sentence = sentence.lower()                     # toàn bộ về thường
    sentence = sentence[0].upper() + sentence[1:]   # viết hoa chữ đầu
    return sentence


```

```python
df['content'] = df['content'].apply(clean_text)

df['content'] = df['content'].apply(sentence_segmentation)

df['content'] = df['content'].apply(
    lambda sentences: [normalize_sentence_case(s) for s in sentences]
)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("VietAI/vit5-base")

df['content'] = df['content'].astype(str)
# Tokenize trực tiếp trên df['content']

df['tokens'] = df['content'].apply(lambda x: tokenizer.tokenize(x))

df['input_ids'] = df['content'].apply(lambda x: tokenizer.encode(x, add_special_tokens=True))
```
